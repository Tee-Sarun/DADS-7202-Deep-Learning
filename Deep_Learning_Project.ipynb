```python
# อ่านไฟล์ HOMEWORK_GUIDELINE.pdf
# เนื่องจากเป็นไฟล์ PDF ผมจะใช้ไลบรารี PyPDF2 เพื่ออ่านข้อความจากไฟล์
# หากยังไม่มี PyPDF2 ให้ติดตั้งก่อน
!pip install PyPDF2

import PyPDF2

pdf_file_path = '/content/HOMEWORK_GUIDELINE.pdf'

# ตรวจสอบว่าไฟล์มีอยู่จริงหรือไม่
try:
    with open(pdf_file_path, 'rb') as file:
        reader = PyPDF2.PdfReader(file)
        text = ''
        for page_num in range(len(reader.pages)):
            page = reader.pages[page_num]
            text += page.extract_text()

    print("อ่านไฟล์ HOMEWORK_GUIDELINE.pdf สำเร็จแล้ว")
    print("เนื้อหาบางส่วน:")
    print(text[:500]) # แสดง 500 ตัวอักษรแรกเพื่อตรวจสอบ

except FileNotFoundError:
    print(f"ไม่พบไฟล์ที่: {pdf_file_path}")
except Exception as e:
    print(f"เกิดข้อผิดพลาดในการอ่านไฟล์ PDF: {e}")
```
